```python
check_rows = []

for _, mapping in semantic_concept_df.iterrows():
    table_id = TABLE_IDS[mapping["table"]]
    column_id = COLUMN_IDS[(mapping["table"], mapping["column"])]

    table_obj = client.get_table(DB_ID, table_id)

    col_obj = next(
        c for c in table_obj.columns
        if c.id == column_id
    )

    current_uri = getattr(col_obj, "concept_uri", None)

    check_rows.append({
        "table": mapping["table"],
        "column": mapping["column"],
        "expected_uri": mapping["concept_uri"],
        "current_uri": current_uri,
        "already_updated": current_uri == mapping["concept_uri"]
    })

semantic_check_before_df = pd.DataFrame(check_rows)

semantic_check_before_df
```

```python
semantic_upload_results = []

for _, row in semantic_check_before_df.iterrows():

    if row["already_updated"] == True:
        semantic_upload_results.append({
            "table": row["table"],
            "column": row["column"],
            "concept_uri": row["expected_uri"],
            "status": "already_updated",
            "error": None
        })
        continue

    try:
        table_id = TABLE_IDS[row["table"]]
        column_id = COLUMN_IDS[(row["table"], row["column"])]

        client.update_table_column(
            database_id=DB_ID,
            table_id=table_id,
            column_id=column_id,
            concept_uri=row["expected_uri"]
        )

        semantic_upload_results.append({
            "table": row["table"],
            "column": row["column"],
            "concept_uri": row["expected_uri"],
            "status": "success",
            "error": None
        })

    except Exception as e:
        semantic_upload_results.append({
            "table": row["table"],
            "column": row["column"],
            "concept_uri": row["expected_uri"],
            "status": "failed",
            "error": repr(e)
        })

semantic_upload_df = pd.DataFrame(semantic_upload_results)

semantic_upload_df
```
